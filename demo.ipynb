# MedTranslate Guard demo

**Prototype, not medical advice.** "No red flag" never means "safe". Always confirm medicine instructions with a pharmacist or health worker.

Run the cells in order. The first cell downloads the repo and installs everything, so this works from a fresh Colab runtime. The first check downloads the translation model (a few minutes; a GPU helps).

```python
import os, subprocess
if not os.path.exists('medtranslate_guard.py'):
    subprocess.run(['git', 'clone', '-q', 'https://github.com/wabix2/medtranslate-guard.git'], check=True)
    os.chdir('medtranslate-guard')
subprocess.run(['pip', 'install', '-q', 'transformers', 'sentencepiece', 'sacrebleu',
                'sentence-transformers', 'scikit-learn', 'pandas', 'numpy', 'matplotlib'], check=True)
print('Ready in', os.getcwd())
```

```python
import medtranslate_guard as mg

out = mg.check_instruction('Take one lozenge every two hours.', lang='amharic')
if out is not None:
    print(out)
```

```python
# Try your own sentence (lang can be 'amharic' or 'oromo')
out = mg.check_instruction('Do not use this ointment near your eyes.', lang='oromo')
if out is not None:
    print(out)
```
